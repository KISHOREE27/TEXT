```python
# ============================================================
#             NLP TEXT ANALYZER - GOOGLE COLAB
#      Modern UI + Tokenization + Stemming + Lemmatization
#             Named Entity Recognition + Sentiment
# ============================================================

# 1. INSTALL LIBRARIES
# Run this notebook cell in Google Colab.
!pip -q install nltk spacy vaderSentiment gradio

# Download required NLTK resources
import nltk
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# Download spaCy English model
!python -m spacy download en_core_web_sm -q


# 2. IMPORTS
import spacy
import gradio as gr
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer


# 3. INITIALIZE NLP TOOLS
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
sentiment_analyzer = SentimentIntensityAnalyzer()
nlp = spacy.load("en_core_web_sm")


# 4. TEXT ANALYSIS FUNCTION
def analyze_text(text):
    if not text or not text.strip():
        message = "Please enter some text to analyze."
        return message, message, message, message, message, message

    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    sentence_result = "\n".join(
        f"{index + 1}. {sentence}" for index, sentence in enumerate(sentences)
    ) or "No sentences found."

    word_result = "\n".join(
        f"{index + 1}. {word}" for index, word in enumerate(words)
    ) or "No words found."

    stemming_result = "\n".join(
        f"{word}  →  {stemmer.stem(word)}"
        for word in words if word.isalpha()
    ) or "No alphabetic words found."

    lemmatization_result = "\n".join(
        f"{word}  →  {lemmatizer.lemmatize(word)}"
        for word in words if word.isalpha()
    ) or "No alphabetic words found."

    doc = nlp(text)
    ner_result = "\n".join(
        f"{entity.text}  →  {entity.label_}"
        for entity in doc.ents
    ) if doc.ents else "No named entities found."

    scores = sentiment_analyzer.polarity_scores(text)
    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive 😊"
    elif compound <= -0.05:
        sentiment = "Negative 😞"
    else:
        sentiment = "Neutral 😐"

    sentiment_result = (
        f"Overall Sentiment: {sentiment}\n\n"
        f"Positive Score : {scores['pos']:.3f}\n"
        f"Negative Score : {scores['neg']:.3f}\n"
        f"Neutral Score  : {scores['neu']:.3f}\n"
        f"Compound Score : {compound:.3f}"
    )

    return (
        sentence_result,
        word_result,
        stemming_result,
        lemmatization_result,
        ner_result,
        sentiment_result
    )


# 5. MODERN GRADIO USER INTERFACE
custom_css = """
.gradio-container {
    max-width: 1180px !important;
    margin: auto !important;
    background: #0b1020 !important;
    color: #e8edff !important;
}
#app-header {
    text-align: center;
    padding: 28px 20px 22px;
    border: 1px solid #293653;
    border-radius: 22px;
    background: linear-gradient(135deg, #151e38 0%, #11172b 55%, #202044 100%);
    margin-bottom: 18px;
}
#app-header h1 {
    font-size: 2.25rem;
    margin: 0 0 8px;
    color: #f4f7ff;
}
#app-header p {
    color: #aab8d8;
    margin: 0;
    font-size: 1rem;
}
#input-panel {
    border: 1px solid #2a3856;
    border-radius: 18px;
    padding: 18px;
    background: #111a2d;
}
#analyze-btn {
    background: linear-gradient(90deg, #6366f1, #8b5cf6) !important;
    border: 0 !important;
    color: white !important;
    font-weight: 700 !important;
}
#clear-btn {
    border: 1px solid #394967 !important;
    background: #18233a !important;
}
.output-box textarea {
    background: #0e1728 !important;
    color: #e2e8ff !important;
    border: 1px solid #2b3b5b !important;
}
label, .tab-nav button {
    color: #dbe5ff !important;
}
footer { display: none !important; }
"""

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="indigo",
        secondary_hue="violet",
        neutral_hue="slate"
    ),
    css=custom_css,
    title="NLP Text Analyzer"
) as demo:

    gr.HTML("""
    <div id="app-header">
        <h1>🧠 NLP Text Analyzer</h1>
        <p>Explore text with six natural language processing tools</p>
    </div>
    """)

    with gr.Group(elem_id="input-panel"):
        gr.Markdown("### ✍️ Input Text")
        text_input = gr.Textbox(
            label="Your text",
            placeholder="Type or paste a paragraph here to begin analysis...",
            lines=6
        )
        with gr.Row():
            analyze_button = gr.Button("✨ Analyze Text", variant="primary", elem_id="analyze-btn")
            clear_button = gr.ClearButton(
                components=[text_input],
                value="↺ Clear Text",
                elem_id="clear-btn"
            )

    gr.Markdown("## 📊 Analysis Workspace")

    with gr.Tabs():
        with gr.Tab("📝 Tokenization"):
            with gr.Row():
                sentence_output = gr.Textbox(
                    label="Sentence Tokenization",
                    lines=12,
                    interactive=False,
                    elem_classes=["output-box"]
                )
                word_output = gr.Textbox(
                    label="Word Tokenization",
                    lines=12,
                    interactive=False,
                    elem_classes=["output-box"]
                )

        with gr.Tab("🌱 Word Forms"):
            with gr.Row():
                stemming_output = gr.Textbox(
                    label="Stemming (Porter Stemmer)",
                    lines=14,
                    interactive=False,
                    elem_classes=["output-box"]
                )
                lemmatization_output = gr.Textbox(
                    label="Lemmatization (WordNet)",
                    lines=14,
                    interactive=False,
                    elem_classes=["output-box"]
                )

        with gr.Tab("🔎 Entities & Sentiment"):
            with gr.Row():
                ner_output = gr.Textbox(
                    label="Named Entity Recognition (spaCy)",
                    lines=12,
                    interactive=False,
                    elem_classes=["output-box"]
                )
                sentiment_output = gr.Textbox(
                    label="Sentiment Analysis (VADER)",
                    lines=12,
                    interactive=False,
                    elem_classes=["output-box"]
                )

    gr.Markdown(
        "<center><small>Powered by NLTK · spaCy · VADER · Gradio</small></center>"
    )

    outputs = [
        sentence_output,
        word_output,
        stemming_output,
        lemmatization_output,
        ner_output,
        sentiment_output
    ]

    analyze_button.click(
        fn=analyze_text,
        inputs=text_input,
        outputs=outputs
    )

    text_input.submit(
        fn=analyze_text,
        inputs=text_input,
        outputs=outputs
    )


# 6. LAUNCH IN COLAB
demo.queue().launch(share=True, debug=True)
```
